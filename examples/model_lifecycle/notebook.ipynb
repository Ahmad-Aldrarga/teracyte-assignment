# Supplied model-development example

Run the same CPU training/evaluation helpers as the CLI. Synthetic regression illustrates the model lifecycle; it is not the camera-image algorithm. Publishing, approval and production selection are candidate tasks.

```python
from pathlib import Path
import sys

root = next(p for p in (Path.cwd(), *Path.cwd().parents)
            if (p / 'examples/model_lifecycle/train.py').is_file())
if str(root) not in sys.path:
    sys.path.insert(0, str(root))
from examples.model_lifecycle.train import run_experiment, load_model, make_dataset, predict, evaluate

output_dir = root / 'data/model-lifecycle'
```

```python
report = run_experiment(output_dir)
report  # Compare held-out MSE with the training-mean baseline.
```

```python
model = load_model(output_dir / 'model.json')
_, _, test_x, test_y = make_dataset()
assert evaluate(model, test_x, test_y) == report['test_mse']
predict(model, test_x[:3])
```
